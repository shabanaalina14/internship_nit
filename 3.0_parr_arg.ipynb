# parameter and arguments

#### variable which is defined inside fun heading is called formal parameters, used to store values which is comming from fun call.

#### variable which is defined inside fun call is called local parameters, used to store temp value.

#### value/ variable which is defined inside fun call called arguments, all arguments pass to the parameters.

## python has 5 types of arguments / parameters

- 1. positional

- 2. default

- 3. positional

- 4. variable length

- 5. variable length arguments(kwargs)


# 1. positional 
Number of arguments in fun call must be equal to number of parameters in fun heading.

```python
def details(id, name, rollno):
    print('{} {} {}'.format(id,name,rollno))
details(1,'alina',54)
details(2,'naresh',65)
details(3,'suresh',87)
```

# 2.default 
when there is common parameter for all fun call then we used default parameter/ arguments.


```python
def studinfo(rollno, name, course='data science'):
    # default value for course is 'data science'
    print('{} {} {}'.format(rollno,name,course))
studinfo(10,'alina')
studinfo(12,'naresh')
studinfo(34,'suresh')
```

```python
def studinfo(rollno, name, course='data science'):
    print('{} {} {}'.format(rollno,name,course))
studinfo(10,'alina','python')
studinfo(12,'naresh')
studinfo(34,'suresh')
```

# 3.positional
some times a end user know names of parameter and arguments but could not able to recall position , in this case we use positional parameter to call parameter according to name.



```python
def digit(a,b,c,d,e,f):
    print(a,b,c,d,e,f)
digit(a=12,b=23,c=34,d=24,e=8)
digit(f=3)
digit(a=12,b=34,c=32,d=43,e=45,f=34)
```

```python
def digit(a,b,c,d,e,f):
    print(a,b,c,d,e,f)
digit(a=12,b=23,c=34,d=24,e=3,f=9)
digit(a=12,b=34,c=32,d=43,e=45,f=34)
```

# variable length 
variable length argumets / paremeter are used to store single formal parameter in fun definition and call many fun from fun call whose type is tuple wit symbol called astrick(*param).

```python
def numbers(*args):
    print(args)
numbers(1,2,3,4,5,6)    
numbers(10,20,30,40,50,60,70)
numbers(100,200,300,400,500,600,700,800,900)
```

```python
def details(sno,name,age,**kwargs):
    print(sno,name,age)
    for key,value in kwargs.items():
        print(key,value)
details(1,'alina',22,course='data science',location='hyderabad')
details(2,'naresh',23,course='data science',location='hyderabad')
details(3,'suresh',24,course='data science',location='hyderabad')
```

```python
# progam for variable length arguments with loop
def details(sno,name,*vals):
    print('serial numbers:{}'.format(sno))
    print('name {}:'.format(name))
    print('values:{}'.format(vals))
    s=0
    for val in vals:
        print('value={}'.format(val))
        s=s+val
        print('sum ={}'.format(s))
details(1,'alina',134,45,35,23,24)
details(3,'naresh',234,34)
details(3,'aliaa',10,20)
```
