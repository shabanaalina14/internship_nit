# String

## Escape characters
An escape character is created by typing a backslash `\` followed by the character you
want to insert.

## Escape character Prints as
`\'` Single quote

`\"` Double quote

`\t` Tab

`\n` Newline (line break)

`\\` Backslash

`\b` Backspace

`\ooo` Octal value

`\r` Carriage Return


```python
# print value in tab
print('\ta\tb')
```

```python
# print value in new line
print('hello \nWorld')
```

```python
#print backslash
print('hello \\ World')
```

```python
#backspace
print('hello\b world')
```

```python
# carriage return
print('hello\r world')
```

### Removing whitespace with strip(), rstrip(), and lstrip()
>>> spam = ' Hello World '
>>> spam.strip()
# 'Hello World'
>>> spam.lstrip()
# 'Hello World '
>>> spam.rstrip()
# ' Hello World'
>>> spam = 'SpamSpamBaconSpamEggsSpamSpam'
>>> spam.strip('ampS')
# 'BaconSpamEggs'


```python
k=' hello world'
print(k.strip())
```

```python
g='   hello world'
print(g.lstrip())
```

```python
g='   hello world'
print(g.rstrip())
```
