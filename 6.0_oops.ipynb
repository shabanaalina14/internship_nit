# encapsulation 

```python
class car():
    def __init__(self):
        self.mil = 10 
        self.com = 'bmw'
        self.price = '1cr'
        
c1 = car()
c2 = car()

print(c1.com, c1.mil,c1.price)
print(c2.com, c2.mil, c2.price)
```

```python
class car:
    def __init__(self):
        self.mil = 10
        self.com = 'bmw'
      
c1 = car()
c2 = car()

c2.mil = 8

c1.mil = 20
c1.com = 'Maruti'

print(c1.com, c1.mil)
print(c2.com, c2.mil)
```
