# Python Variable
A variable is a name refers to the memory location where a value is stored.

A variable is container for data.

# Rules to defining a variable in python.
- Must start with a letter (a-z, A-Z) or an underscore (_).

- Can contain letters, digits (0-9), and underscores, but cannot start with a digit.

- Case-sensitive: age, Age, and AGE are different variables.

- Cannot be a Python keyword (e.g., True, False, if, for, class, etc.).

## syntax

### variable_name = Value


## Examples

```python
# example 1

a=10  # here a is variable storing value 10
print(a)
```

```python
# example 2
a44=10  # here a44 is variable storing value 10
print(a44)
```

```python
# example 3
a_44=10  # here a_44 is variable storing value 10
print(a_44)
```

```python
# example 4
a44_ = 10  # here a44_ is variable storing value 10
print(a44_)
```

```python
# example 5
aZZ = 10  # here aZZis variable storing value 10
print(aZZ)
```

```python
# example 6
age=21
print(age)  # here age is variable storing value 21
```

```python
# example 7
AGE=21
print(AGE)  # here AGE is variable storing value 21

# note : python is case sensitive language, so age and AGE are different variables
```

# Task

```python
# 1.storing and printing value

x=10  # here x is variable storing value 10
print(x)  # here x is variable storing value 10
```

```python
# 2.using variable in expression

x=10  # here x is variable storing value 10
y=20  # here y is variable storing value 20
z=x+y  # here z is variable storing value of x+y=10+20=30
print(z)  # here z is variable storing value 30
```

```python
# 3.changing value of variable

x=10  
print(x) # here x is variable storing value 10
x=20
print(x)  # here x is variable storing value 20

```

```python
# 4.Concatenating string with variable

first_name="Ratan"  # here first_name is variable storing value "Gamesh"
last_name="TATA"  # here last_name is variable storing value "Kumar"
full_name=first_name+" "+last_name  # here full_name is variable storing value of first_name+last_name="Ratan"+" "+"TATA"
print(full_name)  # here full_name is variable storing value "Ratan TATA"
```

```python
# 5.using variable in calculation
length=10  # here length is variable storing value 10
width=5  # here width is variable storing value 5
area=length*width  # here area is variable storing value of length*width=10*5=50
print(area)  # here area is variable storing value 50
```

## Final Notes:
- Variables can store any type of data, including numbers, strings, lists, dictionaries, objects, etc.

- Variables are dynamic, meaning their type is automatically determined by the value assigned to them, and they can change type during execution.
